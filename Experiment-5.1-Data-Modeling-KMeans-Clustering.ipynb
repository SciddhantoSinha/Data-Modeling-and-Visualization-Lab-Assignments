```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("D:\MIT\Semester-6\Data Modelling & Visualization Lab\Excel Files\Customers.csv")
```

```python
X = df.iloc[:, [3,4]].values
from sklearn.cluster import KMeans
wcss = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, init = 'k-means++', random_state = 0)
    kmeans. fit(X)
    wcss.append (kmeans.inertia_) # sum of squar dist of samples to thier closet cluster center
plt. plot(range(1,11), wcss)
plt. title('The elbow method' )
plt.xlabel("no of clusters")
plt.ylabel("wcss")
plt.show()
```

```python
# Implementing the Means Algorithm with the optimal no.of Clusters
kmeans = KMeans(n_clusters = 5, init = 'k-means++', n_init = 10, max_iter = 300, random_state = 0)
y_kmeans = kmeans.fit_predict(X)
```

```python
# Plotting the results	
plt.scatter(X[y_kmeans	==	0, 0],	X[y_kmeans	==	0,	1],	s = 100,	c = 'red', label = 'Clusterl')
plt.scatter(X[y_kmeans	==	1, 0],	X[y_kmeans	==	1,	1],	s = 100,	c = 'blue', label = 'Cluster2')
plt.scatter(X[y_kmeans	==	2, 0],	X[y_kmeans	==	2,	1],	s = 100,	c = 'green', label = 'Cluster3')
plt.scatter(X[y_kmeans	==	3, 0],	X[y_kmeans	==	3,	1],	s = 100,	c = 'cyan', label = 'Cluster4')
plt.scatter(X[y_kmeans	==	4, 0],	X[y_kmeans	==	4,	1],	s = 100,	c = 'magenta', label = 'Cluster5')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 300, c = 'Yellow', label = 'Ce')
plt.xlabel('Annual Income (k$)') 
plt.ylabel('Spending Score (1-100)')
plt.legend() 
plt.show()
```
